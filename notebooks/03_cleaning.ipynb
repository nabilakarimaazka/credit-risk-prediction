Actions:
1. Remove outliers (tidak mungkin seorang peminjam berumur > 80 tahun)
2. Batas nilai lama masa kerja employee

```python
import pandas as pd 

df = pd.read_csv("../data/raw/credit_risk_dataset.csv")
```

```python
df_clean = df.copy()

# Remove usia yang tidak wajar (dataset punya usia 144 tahun)
print(f"Rows before age filter: {len(df_clean):,}")
df_clean = df_clean[df_clean["person_age"] <= 80]
print(f"Rows after removing age >80: {len(df_clean):,}")
```

```python
print(f"Rows before capping length employee filter: {len(df_clean):,}")
df_clean = df_clean[df_clean["person_emp_length"] <= 60]
print(f"Rows after capping length employee >60: {len(df_clean):,}")
```

```python
print(f"\nFinal Cleaned dataset: {df_clean.shape[0]:,} rows x {df_clean.shape[1]} columns")
print("✅ Preprocessing complete")
```

```python
df_clean.to_parquet("../data/processed/credit_risk_dataset_clean.parquet", engine="pyarrow", compression="snappy")
```
